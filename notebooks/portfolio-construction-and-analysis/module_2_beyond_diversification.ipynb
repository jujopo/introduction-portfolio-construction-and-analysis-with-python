## Module 3. Beyond Diversification

Diversification allows you to most efficiently harvest risk premia by eliminating unrewarded risk, specific risk in investors portfolios. It is okay to hold risk if it is rewarded but it is useless to hold unrewarded risk. 

Diversification fails when you need it the most (e.g. 2008). Trying to address this limitation by "improving" diversification simply won't work.

Almost by definition one cannot diversify away systematic risk. You can only diversify specific (ideosyncratic) risk.

__Benefits of insurance?__

Get downside protection while allowing for upside potential. Insurance is dynamic hedging
